```python
import json
import re

exp_report_path = "../../data/reports/cosinesimilarityandgolduris/4ac4ef96-fd26-471f-8ba9-1e796980add5.json"

with open(exp_report_path, 'r', encoding='utf-8') as file:
	report_data = json.load(file)

print(report_data['metadata']['description'])
print('')

def has_missing_angle_brackets_error(sparql_query: str) -> bool:
    regex = r'<https?:\/\/[^>]+>'

    if re.search(regex, sparql_query):
        return False
    else:
        return True

missing_angle_brackets_error_counter = 0
syntax_error_counter = 0

n_rounds = report_data['metadata']['rounds_per_query']

for round in range(n_rounds):
    for query_generation in report_data['data']:
        for generated_queries in query_generation['Generated Queries']:
            if generated_queries['Round'] == round and generated_queries['Has SparQL Parsing Error'] == True:
                syntax_error_counter += 1
                if has_missing_angle_brackets_error(generated_queries['SparQL Query']): missing_angle_brackets_error_counter += 1
    print(f'-----------------------Number of syntax erros in round {round}: {syntax_error_counter}')
    print(f'Number of missing angle bracket syntax errors in round {round}: {missing_angle_brackets_error_counter}')
    syntax_error_counter = 0
    missing_angle_brackets_error_counter = 0
    print("")
```
