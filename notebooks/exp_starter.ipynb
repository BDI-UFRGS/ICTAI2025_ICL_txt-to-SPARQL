```python
from benchmark import SupportedBenchmarks, Benchmark
from endpoint_utils import Endpoint, SupportedEndpoints
from experiment import start_experiment
from nl_query import SupportedEmbedders
from llm_connector import SupportedModels
from prompting_utils import SupportedTemplates
from example_selection import SupportedExampleSelections
from tqdm import tqdm
import pickle
```

# LcQuAD Experiments

## Load the benchmark data

```python
lc_quad = Benchmark(SupportedBenchmarks.LCQUAD)

gold_queries = list(lc_quad.load_gold_queries())
example_queries = list(lc_quad.load_example_queries())
```

## Get the endpoint output to gold and example queries

```python
dbpedia = Endpoint(SupportedEndpoints.DBPEDIA_2016_04)

for gold_query in tqdm(gold_queries, desc=f"Getting the endpoint output for each gold query", leave=False):
    if gold_query.endpoint_output == None:
        gold_query.endpoint_output = dbpedia.get_endpoint_output(gold_query)
        with open(f'../data/persistence/LCQUAD/gold_queries/{gold_query.id}.pkl', 'wb') as dump_file:
            pickle.dump(gold_query, dump_file)

for example_query in tqdm(example_queries, desc=f"Getting the endpoint output for each example query", leave=False):
    if example_query.endpoint_output == None:
        example_query.endpoint_output = dbpedia.get_endpoint_output(example_query)
        with open(f'../data/persistence/LCQUAD/example_queries/{example_query.id}.pkl', 'wb') as dump_file:
            pickle.dump(example_query, dump_file)
```

## Starting experiments

```python
exp = start_experiment(
    description = '1 example, 5 rounds, Bert Embedding Similarity Selection, model LLAMA3_70B_8192, template A',
    model = SupportedModels.LLAMA3_70B_8192,
    prompt_template = SupportedTemplates.A,
    selection_type = SupportedExampleSelections.EMBEDDING_SIMILARITY,
    embedder = SupportedEmbedders.BERT,
    n_rounds = 5,
    n_examples = 1,
    gold_queries = gold_queries,
    example_list = example_queries
)

with open(f'../data/persistence/LCQUAD/experiments/{exp.id}.pkl', 'wb') as dump_file:
    pickle.dump(exp, dump_file)
```
